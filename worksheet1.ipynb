```
#Problem-3 Array Operations
import numpy as np

# Given arrays and vectors
x = np.array([[1, 2],
              [3, 5]])

y = np.array([[5, 6],
              [7, 8]])

v = np.array([9, 10])
w = np.array([11, 12])

#----------------------------------------------------
# Task 1: Matrix Addition (x + y)
#----------------------------------------------------
print("\nTask 1 Answers:")
add = x + y  # element-wise addition
print(add)

#----------------------------------------------------
# Task 2: Matrix Subtraction (x - y)
#----------------------------------------------------
print("\nTask 2 Answers:")
sub = x - y  # element-wise subtraction
print(sub)

#----------------------------------------------------
# Task 3: Element-wise Multiplication (x * y)
#----------------------------------------------------
print("\nTask 3 Answers:")
m = x * y  # multiplication element by element
print(m)

#----------------------------------------------------
# Task 4: Square each element of x
#----------------------------------------------------
print("\nTask 4 Answers:")
sq = x ** 2  # square of each element
print(sq)

#----------------------------------------------------
# Task 5: Dot products
#   (a) v . w
#   (b) x . v
#   (c) x . y
#----------------------------------------------------
print("\nTask 5 Answers:")

# Dot product between vectors v and w
dot = np.dot(v, w)
print("Dot product v . w =", dot)

# Dot product between matrix x and vector v
dot = np.dot(x, v)
print("Dot product x . v =", dot)

# Dot product between matrices x and y
dot = np.dot(x, y)
print("Dot product x . y =\n", dot)

#----------------------------------------------------
# Task 6: Concatenate matrices x and y vertically
#----------------------------------------------------
print("\nTask 6 Answers:")
concat = np.concatenate((x, y), axis=0)
print(concat)

#----------------------------------------------------
# Task 7: Concatenate vectors v and w
#----------------------------------------------------
print("\nTask 7 Answers:")
concat = np.concatenate((v, w))
print(concat)






```

```
#Problme-4 Matrix Operations
import numpy as np

# Given matrices
A = np.array([[3, 4],
              [7, 8]])

B = np.array([[5, 3],
              [2, 1]])

# -----------------------------------------------------
# 1. PROVE A * A^-1 = I
# -----------------------------------------------------
print("\nTask 1: Prove A * A^-1 = I")

A_inv = np.linalg.inv(A)
identity = np.dot(A, A_inv)
print("A * A^-1 =\n", identity)


# -----------------------------------------------------
# 2. PROVE AB ≠ BA
# -----------------------------------------------------
print("\nTask 2: Prove AB ≠ BA")

AB = np.dot(A, B)
BA = np.dot(B, A)

print("AB =\n", AB)
print("BA =\n", BA)
print("Is AB equal to BA? ->", np.array_equal(AB, BA))


# -----------------------------------------------------
# 3. PROVE (AB)^T = B^T A^T
# -----------------------------------------------------
print("\nTask 3: Prove (AB)^T = B^T A^T")

AB_T = np.dot(A, B).T
BT_AT = np.dot(B.T, A.T)

print("(AB)^T =\n", AB_T)
print("B^T A^T =\n", BT_AT)
print("Is (AB)^T equal to B^T A^T? ->", np.allclose(AB_T, BT_AT))


# -----------------------------------------------------
# 4. SOLVE SYSTEM OF LINEAR EQUATIONS (Inverse Method)
# -----------------------------------------------------
print("\nTask 4: Solve system using inverse")

# System of equations:
# 2x - 3y + z = -1
# x - y + 2z = -3
# 3x + y - z = 9

A_eq = np.array([[2, -3, 1],
                 [1, -1, 2],
                 [3, 1, -1]])

B_eq = np.array([-1, -3, 9])

# Solve X = A^-1 B
X = np.dot(np.linalg.inv(A_eq), B_eq)

print("Solution [x, y, z] =", X)







```
