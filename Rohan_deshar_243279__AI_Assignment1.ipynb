```python
import matplotlib.pyplot as plt

# -----------------------------
# Outlier detection function
# -----------------------------
def outliers_detector(data, col):
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return data[(data[col] < lower) | (data[col] > upper)]

# -----------------------------
# Drop missing values (important)
# -----------------------------
sa_df_clean = sa_df.dropna(subset=['hdi', 'gross_inc_percap'])

# -----------------------------
# Detect outliers
# -----------------------------
outliers_hdi = outliers_detector(sa_df_clean, 'hdi')
outliers_gni = outliers_detector(sa_df_clean, 'gross_inc_percap')

# -----------------------------
# Print results
# -----------------------------
print(f"Number of HDI Outliers: {len(outliers_hdi)}")
print(f"Number of GNI Outliers: {len(outliers_gni)}")

print(
    outliers_gni[['country', 'year', 'gross_inc_percap']]
    .sort_values(by='gross_inc_percap', ascending=False)
)

# -----------------------------
# Plot: GNI vs HDI with outliers
# -----------------------------
plt.figure(figsize=(10, 6))

# Normal observations
plt.scatter(
    sa_df_clean['gross_inc_percap'],
    sa_df_clean['hdi'],
    alpha=0.6,
    label='Normal'
)

# GNI outliers
plt.scatter(
    outliers_gni['gross_inc_percap'],
    outliers_gni['hdi'],
    label='GNI Outliers'
)

plt.title('Outlier Detection: GNI vs HDI (South Asia – All Years)')
plt.xlabel('Gross National Income per Capita')
plt.ylabel('HDI')
plt.legend()
plt.show()

```

```python
# 7) Outlier Detection Function (1.5 × IQR)
def detect_outliers(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    outliers = df[(df[col] < Q1 - 1.5*IQR) | (df[col] > Q3 + 1.5*IQR)]
    print(f"{len(outliers)} outliers detected in column '{col}'")
    return outliers
```

```python
# Detect HDI Outliers

# South Asia
sa_outliers = detect_outliers(sa_2020_22, 'hdi')
print("South Asia HDI Outliers:")
print(sa_outliers[['country', 'year', 'hdi']])

# Middle East
me_outliers = detect_outliers(me_2020_22, 'hdi')
print("\nMiddle East HDI Outliers:")
print(me_outliers[['country', 'year', 'hdi']])
```
